```python
import numpy as np
def Func(VarInd):
    return VarInd**6 - VarInd**2 * np.sin(2 * VarInd)
```

```python
def PuntosPesos(N):
    x, w = np.polynomial.legendre.leggauss(N)
    
    return x, w
    
def EscalarIntervalo(LimInf, LimSup, x, w):
    x = 0.5 * (LimSup - LimInf) * x + 0.5 * (LimSup + LimInf)
    w = 0.5 * (LimSup - LimInf) * w
    
    return x, w
```

```python
def CuadraturaGaussiana(LimInf, LimSup, N):
    
    x, w = PuntosPesos(N)
    
    x, w = EscalarIntervalo(LimInf, LimSup, x, w)
    
    Integral = 0
    
    for i in range(N):
        Integral += w[i] * Func(x[i])
    
    return Integral
```

```python
LimInf = 1
LimSup = 3
N = 3

Resultado = CuadraturaGaussiana(LimInf, LimSup, N)

print(Resultado)
```

```python
for N in range(1, 11):
    Resultado = CuadraturaGaussiana(1, 3, N)
    print("N =", N, "Resultado =", Resultado)
```
